```python
size_list = range(1,801)
mean_sample=[]
std_sample=[]
for i in size_list:
    mean_sample.append(pokemon["HP"].sample(i).mean())
    std_sample.append(pokemon["HP"].sample(i).std())
sample_simulation = pd.DataFrame({"Sample_size":size_list,"sample_mean":mean_sample,"sample_sd":std_sample})
sample_simulation
```

```python
plt.plot(sample_simulation["Sample_size"],sample_simulation["sample_mean"],color="r",alpha=0.5)
plt.plot(sample_simulation["Sample_size"],sample_simulation["sample_sd"],color="b",alpha=0.5)
plt.axhline(pokemon["HP"].mean(),linestyle="dotted",label="Population avg")
plt.axhline(pokemon["HP"].std(),linestyle="dotted",label="Population std")
plt.legend()
plt.show()
```
